## Time Complexity

Use of time xomplexity makes it easy to estimate the running time of a program. Performing an accurate calculation of a program's operation time is a very labour-intensive process. Therefore we will not make an accurate measurement: just a measurement of a certian order of magniturde. Complexity can be viewed as the maximum number of primitive operations tahtt a programme may execute. Regular operations are single additions, multiplications, assignments etc. We may leave some operations uncoutned and concentrate on those that are performed the largest number of times. Such are referred to as **dominant**.

The number of dominant operationds depends on the specific input data. We usually want to know how the performance time depends on a particular aspect of the data. This is most frequently the data size, but it can also be the size of a square matrix or the value of some input variable.



```python
def dominant(n):
    result = 0
    for i in range(n):
        result += 1
    return result
```

More specifically in the case of O(n) the program may perform c x n operations where c is a constant, however it may not perform n^2 operations since this involves a different order of magnitude of data. In other words, when calculating the complexity we omit constants, i.e regardless of whether the loops is executed 20 x n times or n / 5 times we still have complexity O(n).

There is a simple yet powerful technique that allows for the fast calculation of sums of elements in given slice (contiguous segments of array). Its main idea uses prefix sums which are defined as the consecutive totals of the first 0,1,2,...,n elements of an array. 

i.e the partial sums of values

we can easily calculate the prefix sums in O(n) time complexity notice that the total pk equals pk-1 + ak-2 so each consecutive value can be calculated in a constant time. 

```python
# Counting prefix sums - O(n)

def prefix_sums(A):
    n = len(A)
    P = [0] * (n+1)
    for k in range(1,n+1):
        P[k] = P[k-1] + A[k-1]
    return P
```

Similarly, we can calculate suffix sums, which are the totals of the k last values. Using prefix (or suffix) sums allows us to claculate the total of any slice of the array very quickly. Sorting is the process of arranging data in a certain order. Usually, we sort by the value of the elements. We can sort numbers, words, pairs, etc. For example we can sort students by their height and we can sort cities in alphabetical order or by their numbers of citizens. The most-used orders are numerical order and alphabetical order. Let's consider the simplest set, an array consisting of integers:

There are many sorting algorithms, and they differ considerably in terms of their time complexity and use of memory. Here we descrive some of them

#### Selection sort

The idea: Find the minimal element and swap it with the first element of an array. Next just sort the rest of the array, without the first element, in the same way. Notice that after k iterations (repetition of everything inside the loop) the first k elements will be sorted in the right order ( this type of a property is called the loop invariant).

```python
# Selection sort - O(n^2)

def selectionSort(A):
    n = len(A)
    for k in range(n):
        minimal = k
        for j in range(k+1, n):
            if A[j] < A[minimal]:
                minimal = j
        A[k] , A[minimal] = A[minimal], A[k]
    return A

# the time complexity is quadratic


```

#### Counting Sort

The idea: First, count the elements in the array of counters, next just iterate through the array of counters in increasing order.
Notice that we have to know the range of the sorted values. If all the elements are in the set {0,1,...,k} then the array used for counting shoudl be of size k+1. The limitation here may be available memory. 

Counting sort O(n+k)

```python
def countingsort(A,k):
    n = len(A)
    count = [0] * (k+1)
    for i in range(n):
        count[A[i]] += 1
    p = 0
    for i in range(k+1):
        for j in range(count[i]):
            A[p] = i
            p += 1
    return A
    
```

The time complexity here is O(n+k). We need additional memory O(k) to count all the elements. At first sight, the time complexity of the above implementation may appear greater. However, all the operations in lines 9 and 10 are performed not more than O(n) times. 

#### Merge sort
The idea: Divide the unsorted array into two halves, sort each half separately and then just merge them. After the split, each part is halved again.
We repeat this algorithm until we end up with individual elements, which are sorted by definition. The merging of two sorted arrays consisting of k elements takes O(k) time; just repeatedly choose the lower of the first elements of the two merged parts. The length of the array is halved on each iteration. In this way, we get consevutive levels with 1,2,4,8,.. slices. for each level the merging of all consecutive pairs of slices requires O(n) time. The number of levels is O(logn) so the total time complexity is O(nlogn)

## Queue 

The queue is a basic data structure in which new elements are inserted at the back but old elements are removed from the front. The idea of the queue can be illustrated by a line of customers in a grocery store. New people join the back and the next person to be served is the first one in the line 


The queue can be represented by an array for storing the elements. A part of the array we should also remember is the front(head) and the back (tail) of the queue. We must be sure to devlare sufficient space for the array.

```python
#Push Pop Size Empty Function O(1)

queue = [0] * N 
head, tail = 0, 0 
def push(x):
    global tail 
    tail = (tail + 1) % N
    queue[tail] = x
def pop():
    global head
    head = (head +1) % N
    return queue[head]
def size():
    return (tail - head + N) % N
def empty():
    return head == tail
```

The push function adds an element to the queue. The pop function removes and returns an element from the front of the queue ( we shouldn't perform a pop operation on an empty queue). The empty function checks whether the queue is empty. The size fn returns the number of elements in the queue.



## Maximum slice problem

Let's define a problem relating to maximum slices. You are given a sequence of n integers $$a_0 , a_1, a_{n-1}$$ and the task is to find the slice with the largest sum. More precisely, we are looking for two indices p,q such that the total $$a_p + a_{p+1}+ ... + a_q$$ is maximal. We assume that the slice can be empty and its sum equals 0. 


In the picture, the slice with the largest sum is highlighted in grey. The sum of this slice equals 10 and there is no
slice with a larger sum. Notice that the slice we are looking for may contain negative integers, as shown above.


#### Solution with O(n^3) time complexity

The simplest approahc is to analyze all the slices and choose the one with the largest sum.

```python
# Maximal Slice O(n^3)

def slow_max_slice(A):
    n = len(A)
    result = 0 
    for p in range(n):
        for q in range(p,n):
            sum = 0
            for i in range(p,q+1):
                sum += A[i]
            result = max(result, sum)
    return result

# Analyzing all possible slices requires O n squared time complexity and for each of them we compute the 
#total in O n time complexity this is the most straightforward solution however it is far from optimal
```

#### Solution with O(n^2) time complexity

We can easily imporve out last solution. Notice that the prefix sum allows the sum of any slice to be computed in a constant time. With this approach the time complexity reduces to n^2 We assume that pref is an array of prefix sums


```python

def quadratic_max_slice(A, pref):
    n = len(A)
    result = 0
    for p in range(n):
        for q in range(p, n):
            sum = pref[q + 1]- pref[p]
            result = max(result, sum)
    return result
```

We can also solve this problem without using prefix sums, within the same time complexity. Assume that we know the sum of slice (p,q) so $$s = a_p + a_{p+1} + ... + a_q$$

The sum of the slice with one more element (p,q+1) equals $s + a_q+1$. Following this observation, there is no need to compute the sum each time from the beginning, we can use the previously calculated sum.

```python
def quadratic_max_slice(A):
    n = len(A)
    result = 0
    for p in range(n):
        sum = 0 
        for q in range(p, n):
            sum += A[q]
            result = max(result, sum)
    return result
```

## The optimal solution with O(n) time complexity

For each position, we compute the largest sum that ends in that position. 

If we assume that the maximum sum of a slice ending in position i equals max_ending, 

then the maximum slice ending in position i+1 equals max(0, max_ending + $a_i+1$)

```python
# Maximal Slice O(n)

def golden_max_slice(A):
    max_ending = 0
    max_slice = 0
    for a in A:
        max_ending = max(0, max_ending + a)
        max_slice = max(max_slice, max_ending)
    return max_slice

# If adding the current element (a) to the current slice sum (max_ending) makes it negative, reset to zero (start a new slice).
# Keep track of the highest slice sum you have seen so far (max_slice).
    
```

def count_grouped_equal_substrings(s: str) -> int:
    groups = []
    count = 1

    # Count consecutive groups of 0's or 1's
    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            count += 1
        else:
            groups.append(count)
            count = 1
    groups.append(count)

    # Count substrings from adjacent groups
    total = 0
    for i in range(len(groups) - 1):
        total += min(groups[i], groups[i+1])

    return total

## Binary Search algorithm

The binary search is a simple and very useful algorithm whereby many linear algorithms can be optimized to run in logarithmic time.

#### Intuition

Imagine the following game. The computer selects an integer value between 1 and 16 and our goal is to guess this number with a minimum number of questions. For each guessed number the computer states whether the guessed number is equal to, bigger or smaller than the number to be guessed.

The iterative check of all the successive values 1,2,...,16 is linear because with each question the set of the candidates is reduced by one. 

The goal is to ask a question that reduces the set of candidates maximally. The best option is to choose the middle element, as doing so causes the set of candidates to be halved each time. With this approach we ask the logarithmic number of questions at maximum

In a binary search we use the information that all the elements are sorted. 

For every step of the algorithm we should remember the beginnign and the end of the remaining slice. The middle element of the slice can easily be calculated as mid = (beg + end) / 2


```python
def binarySearch(A, x): # A - sorted list , x - target value of search
    n = len(A)
    beg = 0 # beginnign index of search range
    end = n - 1
    result = -1 # initialises the result to -1, stores the index of the last element leq to x
    while (beg <= end): # loop continues as long as search range is valid
        mid = (beg + end) // 2  #floor div
        if (A[mid] <= x): # checks if middle element is leq than target
            beg = mid + 1 # if it is move the beginning of the search to mid +1
            result = mid
        else:  # else shrink search to left of mid
            end = mid - 1
    return result
```

the above algorithm will find the largest element which is less than or equal to x. In subsequent iteration the number of candidates is halved, so the time complexity is O(logn)


#### Binary Search on the result

In many tasks, we should return some integer that is both optimal and that meets certain conditions. We can often find this number using a binary search. We guess some vale and then check whether the result should be smaller or bigger. At the start we have a certain range in which we can find the result. After each attempt the range is halved, so the numebr of questions can be estimated by O(logn)

Thus the problem of finding the optimal value reduces to checking whether some value is valid and optimal. The latter problem is often much simpler and the binary search adds only a log n factor to the overall timef complexity.

## Quick sort
# Recursive algorithm

pick a pivot define array into three different pieces left side all values lest than pivot middle pivot, right strictly greater than pivot.

Then split the segments again




```python
def quick_sort(arr):

#as this is a recursive algorithm we need a base case

    if len(arr) <= 1:
        return arr
    

    p = arr[-1] # use the highest value at the last index

    L = [x for x in arr[:-1] if x <= p] # everything up until last index if x <=p
    R = [x for x in arr[:-1] if x > p ] # everything greater than p

    L = quick_sort(L)
    R = quick_sort(R) # recursion

    return L + [p] + R
    
```

##### Given a string determin if it is a palindrom, considering only alphanumeric characters

```python
def is_palindrome(s:str) -> bool:
    l,r = 0 , len(s) -1
    while l < r:
        while l < r and not s[l].isalnum():
            l += 1
        while l < r and not s[r].isalnum():
            r -= 1
        if s[l].lower() != s[r].lower():
            return False
        l += 1
        r -= 1

    return True
```

#### Sliding Window 

Instead of tracking just two positions you are managing an entire range of values. This range is called your window.

If a problem is asking about contiguous segments like a substring a subarry or a group of continuous elements a sliding window is almost always the best answer.

At any moment you are onlu concerned about what is in the window.

There are two types:

- Fixed window - where the size of the window is predetermined and is constant throughout
- Dynamic window - where the window can shrink and expand as necessary.

Fixed Window:
a) Find the maximum average of any subarray of size k 
b) Return the sum of every k-length block
c) Find the subarray of length k with the largest/smallest X

the windows length never changes, at each step you are adding one new element to the right and removing one element form the left, you update you rresult based on the content of the window


Dynamic window:
a) Find the length of the longest substring with at most K unique characters
b) What's the smallest subarray with a sum greater than a target
c) Return the longest window where a certain rule is valid

grow element right side while true. if the window is invalid start removing elements from the left until it becomes valid again then start appending right



```python
# Find the largest sum among all subarrays of length k in an array
# nbote += for numbers adds and assigns to variable 
# for strings this concatenates to the string
# lists this extends
# for sets |= union
# dictionaries you can update values by d[key] += 1


def subarray_sum_fixed(nums,k):
    window_sum = 0
    for i in range(k): # sum over first k elements
        window_sum += nums[i]
    largest = window_sum


    # sliding window

    for right in range(k, len(nums)):
        left = right - k # 3 - 3 = 0 
        # so remove element at index 0 
        window_sum -= nums[left] 
        window_sum += nums[right]
        largest = max(largest, window_sum)

    return largest
```

```python
# Dynamic window

# Find the length of the longest substring of a given string without repeating characters

def longest_substring_without_repeating_characters(s):
    longest = 0
    l = 0
    # we set up a dictionary called counter using default dict to have count zero
    counter: dict[str, int] = defaultdict(int)
    for r in range(len(s)):
        counter[s[r]] += 1
        while counter[s[r]] > 1:
            counter[s[l]] -= 1
            l += 1
            longest = max(longest, r - l + 1)
    return longest


```

```python
def find_min_rotated(arr: List[int]) -> int:
    left, right = 0, len(arr) - 1
    boundary_index = -1 

    while left <= right:
        mid = (left + right) // 2
        if arr[mid] <= arr[-1]:
            right = mid - 1 
            boundary_index = mid
        else:
            left = mid + 1
    return boundary_index
    
```

```python
find_min_rotated([30,40,50,10,20])
```

# Backtracking

Is a recursive problem solving technique that explores all possible configurations of a solution but efficiently backs up when it realises it wont work

This is used when the solution involves combinations, permutations, building up a partial solution want all possible solutions

## Dynamic Programming 
Four different steps
1) Naive Recursive solution, has very big run time
2) Top-down DP technique Memoization, i.e using a cache to memorise function values, uses recursion
3) Bottom-up DP, tabulation, filling out a table or array using loops
4) Constant Space solution (if possible)


```python
# Fibonacci Number each number is the sum of the two preceding ones
def fib(n:int) -> int:
    # Recursive Solution
    if n == 0:
        return 0
    if n == 1:
        return 1


    return fib(n-2) + fib(n-1)
```

If we have already computed a function call we dont have to do the entire tree again just use the saved value.

we usually implement this in a hash map {f(0):0, f(1):1, f(2):1, f(3):2, ....}

the general term is a cache rememebering things this is usually called **memo** as a object


```python
def fib(n):
    memo = {0:0, 1:1}

    def f(x):
        if x in memo:
            return memo[x]
        else:
            memo[x] = f(x-1) + f(x-2)
            return memo[x]

    return f(n)            
```
